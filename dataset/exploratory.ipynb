# Techbro Twitter Indonesia Dataset - Exploratory

Dataset: https://www.kaggle.com/datasets/israhabibi/techbro-twitter-indonesia

Tiga file:
- `techbro_tweets.csv`  : 1 baris per tweet (granular)
- `techbro_activity.csv`: 1 baris per (hari x user)
- `techbro_topics.csv`  : 1 baris per topik hangat (kolom `days`)

```python
import pandas as pd
import matplotlib.pyplot as plt

tweets   = pd.read_csv("techbro_tweets.csv")
activity = pd.read_csv("techbro_activity.csv")
topics   = pd.read_csv("techbro_topics.csv")

print("tweets  :", tweets.shape)
print("activity:", activity.shape)
print("topics  :", topics.shape)
```

## 1. Tweet per hari

```python
tpd = activity.groupby("day")["tweet_count"].sum().sort_index()
print(tpd.tail(10))
tpd.tail(30).plot(kind="bar", figsize=(12,4), color="#e63946")
plt.title("Tweet techbro per hari")
plt.tight_layout()
plt.show()
```

## 2. Top user paling aktif

```python
top_users = (activity.groupby("user_handle")["tweet_count"]
             .sum().sort_values(ascending=False).head(10))
print(top_users)
top_users.plot(kind="barh", figsize=(8,5), color="#1d3557")
plt.title("Top 10 user paling aktif")
plt.tight_layout()
plt.show()
```

## 3. Topik hangat + hari muncul

```python
for _, r in topics.iterrows():
    print(f"- {r['topic']} ({r['tweet_count']} tweet, hari: {r['days']})")
    print(f"  {r['summary'][:120]}")
```
