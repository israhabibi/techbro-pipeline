Generate offline demo exports first. Install the locked `analysis` extra, then open this notebook from `dataset/`. Demo accounts are fictional.


```python
from pathlib import Path
import os
import pandas as pd
import matplotlib.pyplot as plt
folder = Path(os.environ.get("DATASET_DIR", "."))
tweets = pd.read_csv(folder / "techbro_tweets.csv")
activity = pd.read_csv(folder / "techbro_activity.csv")
topics = pd.read_csv(folder / "techbro_topics.csv")
print({"tweets": len(tweets), "activity": len(activity), "topics": len(topics)})
```

```python
activity.groupby("day")["tweet_count"].sum().sort_index().plot(kind="bar", title="Technology posts per day")
plt.tight_layout()
```
